```python
# fungsi i/o
nama = input("Masukkan nama Anda: ")
umur = input("Masukkan umur Anda: ")
nilai = input("Masukkan nilai Anda: ")
print("Nama Anda adalah " + nama)
print("Umur " + umur + " tahun")
print("Nilai Anda adalah " + nilai)
```

```python
# struktur pemilihan (selection)
nilai = 71
if nilai > 70:
   print("Lulus")
else:
   print("Tidak lulus")
```

```python
# struktur pemilihan dengan fungsi i/o
nama = input("Masukkan nama Anda: ")
nilai = input("Masukkan nilai Anda: ")
if int(nilai) >= 85:
   print(nama + " Grade A")
elif int(nilai) >= 75:
   print(nama + " Grade B")
elif int(nilai) >= 65:
   print(nama + " Grade C")
elif int(nilai) >= 55:
   print(nama + " Grade D")
else:
   print(nama + " tidak lulus")
```

```python
# struktur perulangan (looping) / default dimulai dari 0
# for
for i in range(5):
   print(i)
```

```python
# struktur perulangan (looping) / mulai dari 1
# for
for i in range(1, 6):
   print(i)
```

```python
# while
i = 1
while i <= 5:
   print(i)
   i += 1
```

```python
# while
i = 1
while i < 5:
   i += 1
   print(i)
```

```python
# break
for i in range(10):
  if i == 5:
    break
  print(i)
```

```python
# continue
for i in range(5):
  if i == 2:
    continue
  print(i)
```

```python
# nested loop
for i in range(2):
   for j in range(3):
      print(i, j)
```
