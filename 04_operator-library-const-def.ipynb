```python
# operator dasar
a = 3
b = 2
c = a + b

print(f"Penjumlahan: {a} + {b} =", c)
print(f"Pengurangan: {a} - {b} =", a - b)
print(f"Perkalian: {a} * {b} =", a * b)
print(f"Pembagian: {a} / {b} =", a / b)
print(f"Pembagian Bulat: {a} // {b} =", a // b)
print(f"Sisa Bagi: {a} % {b} =", a % b)
print(f"Pangkat: {a} ** {b} =", a ** b)
```

```python
# library math
import math

nilai = 81
akar = math.sqrt(nilai)
print(f"Akar dari {nilai} adalah {akar}")

nil1 = 7.8
nil2 = 7.2
nil3 = 7.5
print(f"Pembulatan {nil1} menggunakan floor adalah", math.floor(nil1))
print(f"Pembulatan {nil1} menggunakan ceil adalah", math.ceil(nil1))
print(f"Pembulatan {nil2} menggunakan floor adalah", math.floor(nil2))
print(f"Pembulatan {nil2} menggunakan ceil adalah", math.ceil(nil2))
print(f"Pembulatan {nil3} menggunakan floor adalah", math.floor(nil3))
print(f"Pembulatan {nil3} menggunakan ceil adalah", math.ceil(nil3))

print("Nilai faktorial dari 5 adalah", math.factorial(5))
```

```python
# konstanta dari math
import math

print(math.pi)
print(round(math.pi,2))
print(math.e)
print(round(math.e,2))
print(math.tau)
print(round(math.tau,2))
print(math.inf)
print(math.nan)
```

```python
# fungsi custom
def tambah(a, b):
  hasil = a + b
  return hasil

print(tambah(3, 4))

def luas_segitiga(alas, tinggi):
  luas = (alas * tinggi) / 2
  return luas

print(luas_segitiga(3, 4))
```

```python
# volume bola
import math

def volumebola(r):
  volume = (4/3) * math.pi * (r**3)
  return round(volume,2)

volumebola(6)
```

```python
# Aplikasi Konversi Suhu

# Celsius
def celsius_ke_fahrenheit(c):
    fahrenheit = (c * 9/5) + 32
    return round(fahrenheit, 2)

def celsius_ke_reamur(c):
    reamur = c * 4/5
    return round(reamur, 2)

def celsius_ke_kelvin(c):
    kelvin = c + 273.15
    return round(kelvin, 2)

# Fahrenheit
def fahrenheit_ke_celsius(f):
    celsius = (f - 32) * 5/9
    return round(celsius, 2)

def fahrenheit_ke_reamur(f):
    reamur = (f - 32) * 4/9
    return round(reamur, 2)

def fahrenheit_ke_kelvin(f):
    kelvin = (f - 32) * 5/9 + 273.15
    return round(kelvin, 2)

# Reamur
def reamur_ke_celsius(r):
    celsius = r * 5/4
    return round(celsius, 2)

def reamur_ke_fahrenheit(r):
    fahrenheit = (r * 9/4) + 32
    return round(fahrenheit, 2)

def reamur_ke_kelvin(r):
    kelvin = (r * 5/4) + 273.15
    return round(kelvin, 2)

# Kelvin
def kelvin_ke_celsius(k):
    celsius = k - 273.15
    return round(celsius, 2)

def kelvin_ke_fahrenheit(k):
    fahrenheit = (k - 273.15) * 9/5 + 32
    return round(fahrenheit, 2)

def kelvin_ke_reamur(k):
    reamur = (k - 273.15) * 4/5
    return round(reamur, 2)


# Konversi Celsius
print("=== KONVERSI CELSIUS ===")
celsius = 100
print(f"{celsius} Celsius =", celsius_ke_fahrenheit(celsius), "Fahrenheit")
print(f"{celsius} Celsius =", celsius_ke_reamur(celsius), "Reamur")
print(f"{celsius} Celsius =", celsius_ke_kelvin(celsius), "Kelvin")

# Konversi Fahrenheit
print("\n=== KONVERSI FAHRENHEIT ===")
fahrenheit = 212
print(f"{fahrenheit} Fahrenheit =", fahrenheit_ke_celsius(fahrenheit), "Celsius")
print(f"{fahrenheit} Fahrenheit =", fahrenheit_ke_reamur(fahrenheit), "Reamur")
print(f"{fahrenheit} Fahrenheit =", fahrenheit_ke_kelvin(fahrenheit), "Kelvin")

# Konversi Reamur
print("\n=== KONVERSI REAMUR ===")
reamur = 80
print(f"{reamur} Reamur =", reamur_ke_celsius(reamur), "Celsius")
print(f"{reamur} Reamur =", reamur_ke_fahrenheit(reamur), "Fahrenheit")
print(f"{reamur} Reamur =", reamur_ke_kelvin(reamur), "Kelvin")

# Konversi Kelvin
print("\n=== KONVERSI KELVIN ===")
kelvin = 373.15
print(f"{kelvin} Kelvin =", kelvin_ke_celsius(kelvin), "Celsius")
print(f"{kelvin} Kelvin =", kelvin_ke_fahrenheit(kelvin), "Fahrenheit")
print(f"{kelvin} Kelvin =", kelvin_ke_reamur(kelvin), "Reamur")
```
