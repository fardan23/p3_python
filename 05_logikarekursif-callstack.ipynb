```python
# hitung mundur
def hmundur(angka):
  # base case
  if angka <= 0:
    print("selesai.....")
  else:
    print(angka)
    # recursive case
    hmundur(angka-1)
hmundur(5)
```

```python
# hitung deret -> 3 + 2 + 1 = 6
def jderet (angka):
  if angka == 1:
    print(angka)
    return 1
  else:
    print(angka)
    return angka + jderet(angka-1)
print(jderet(3))
```

```python
# input hitung deret -> n + (n-1) + .. + 1 = 6
def jderet(angka):
  if angka == 1:
    print(angka)
    return 1
  else:
    print(angka)
    return angka + jderet(angka-1)

nil = int(input("Input nilai: "))
jderet(nil)
```

```python
# faktorial -> 5! = 5 x 4 x 3 x 2 x 1
def faktorial(angka):
  if angka == 0:
    return 1
  hasil = angka * faktorial(angka-1)
  print(f"faktorial {angka} adalah " , hasil)
  return hasil
faktorial(5)
```

```python
# input faktorial -> n! = n x (n-1) x .. x 1
def faktorial(angka):
  if angka == 0:
    return 1
  hasil = angka * faktorial(angka-1)
  print(f"faktorial {angka} adalah " , hasil)
  return hasil

n = int(input("Input nilai: "))
faktorial(n)
```

```python
# call stack
def box(isibox):
  print("Masukkan isibox: ", isibox)
  if isibox < 3:
    box(isibox + 1)
  else:
    print("selesai")
  print("ambil isibox: ", isibox)
box(1)
```

```python
def cuci_piring(nomor_piring):
  print("Menumpuk piring kotor nomor:", nomor_piring)
  if nomor_piring < 3:
    cuci_piring(nomor_piring + 1)
  else:
    print("Tumpukan piring selesai dibuat, sekarang mulai mencuci")
  print("Selesai mencuci piring nomor:", nomor_piring)
cuci_piring(1)
```
