```python
# Program Manajemen Truk Gudang

KAPASITAS_MAKSIMAL = 5000
total_berat = 0

print("=== MANAJEMEN TRUK GUDANG ===")
print("Kapasitas maksimal:", KAPASITAS_MAKSIMAL, "kg")

while True:
    berat = float(input("\nMasukkan berat barang (kg): "))

    if total_berat + berat > KAPASITAS_MAKSIMAL:
        print("Over-capacity! Barang tidak dapat dimuat.")
        print("Sisa kapasitas:", KAPASITAS_MAKSIMAL - total_berat, "kg")
        continue

    total_berat += berat

    print("Barang berhasil dimuat.")
    print("Total berat sekarang:", total_berat, "kg")

    lanjut = input("Tambah barang lagi? (y/n): ")

    if lanjut.lower() == "n":
        break

if total_berat >= 4000:
    armada = "Heavy Duty Truck"
elif total_berat >= 2000:
    armada = "Medium Duty Truck"
elif total_berat > 0:
    armada = "Light Pick-Up"
else:
    armada = "Tidak ada kendaraan"

print("\n=== HASIL AKHIR ===")
print("Total berat muatan :", total_berat, "kg")
print("Jenis kendaraan    :", armada)
```
