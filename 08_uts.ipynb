```python
# Nomor 2
def input_float():
    while True:
        try:
            angka = float(input("Masukkan angka desimal: "))
        except ValueError:
            print("Peringatan: input harus berupa angka!")
        else:
            print(f"Angka valid: {angka}")
            break
        finally:
            print("Status: satu siklus pemeriksaan selesai\n")
    return angka

hasil = input_float()
```
