```python
# encapsulation
class Hewan:
  def __init__(self, nama, jk): # method constructor
    self.nama = nama # attribute public
    self.jk = jk # attribute public
  def info(self): # method / fungsi public
    return f"{self.nama} adalah hewan dengan jenis kelamin {self.jk}"
  def suara(self): # method / fungsi public
    return "Hewan bersuara..."

# inheritance
class Kucing(Hewan):
  def __init__(self, nama, jk, warna):
    super().__init__(nama, jk)
    self.warna = warna # attribute private
  #polymorphism
  def info(self):
    return f"{super().info()} warna {self.warna}"
  def suara(self):
    return "meow..."

class Kambing(Hewan):
  def __init__(self, nama, jk, ras):
    super().__init__(nama, jk)
    self.ras = ras # attribute private
  def info(self):
    return f"{super().info()} ras {self.ras}"
  def suara(self):
    return "mooo..."

kucing = Hewan("Kucing", "Jantan")
print(kucing.info())
print(kucing.suara())

kucing1 = Kucing("Kucing1", "Jantan", "Putih")
print(kucing1.info())
print(kucing1.suara())

kambing = Kambing("Kambing", "Betina", "Domba")
print(kambing.info())
print(kambing.suara())
```

```python
class Kendaraan:
  def __init__(self, merk, warna):
    self.merk = merk
    self.warna = warna
  def info(self):
    return f"Kendaraan {self.merk}, warna {self.warna}"

class Mobil(Kendaraan):
  def __init__(self, merk, warna, tipe):
    super().__init__(merk, warna)
    self.tipe = tipe
    self._kecepatan = 0
  # setter
  _set_kecepatan = lambda self, k: setattr(self, "_kecepatan", k)
  # _get_kecepatan = lambda self: getattr(self, "_kecepatan", 0)
  def tancap_gas(self, tambah):
    self._kecepatan += tambah
    return f"Mobil {self.merk} melaju dengan kecepatan {self._kecepatan} km/jam"
  def info(self):
    return f"{super().info()}, tipe {self.tipe}"

motor = Kendaraan("Honda", "Merah")
print(motor.info())

mobil = Mobil("Toyota", "Hitam", "Sport")
print(mobil.info())
print(mobil.tancap_gas(100))
print(mobil.tancap_gas(50))
```
