```python
# Program Slip Gaji PT Mulia Jaya

print("=== SLIP GAJI PT MULIA JAYA ===")

kelas = input("Masukkan kelas pegawai (A/B/C): ").upper()

if kelas == "A":
    gaji_pokok = 5_000_000
    tarif_lembur = 50_000
    jam_lembur = 5

elif kelas == "B":
    gaji_pokok = 4_000_000
    tarif_lembur = 35_000
    jam_lembur = 2

elif kelas == "C":
    gaji_pokok = 3_500_000
    tarif_lembur = 20_000
    jam_lembur = 6

else:
    print("Kelas pegawai tidak valid.")
    exit()

uang_lembur = jam_lembur * tarif_lembur

total_gaji = gaji_pokok + uang_lembur

print("\n=== SLIP GAJI ===")
print("Kelas Pegawai  :", kelas)
print("Gaji Pokok     : Rp", format(gaji_pokok, ",").replace(",", "."))
print("Tarif Lembur   : Rp", format(tarif_lembur, ",").replace(",", "."))
print("Jam Lembur     :", jam_lembur, "jam")
print("Uang Lembur    : Rp", format(uang_lembur, ",").replace(",", "."))
print("Total Gaji     : Rp", format(total_gaji, ",").replace(",", "."))
```
