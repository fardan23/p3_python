```python
try:
    angka =  int("sepuluh")
except ValueError:
    print("angka tidak boleh huruf")
else:
    print(f"cetak {angka}")
finally:
    print("eksekusi selesai")
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n-1)

import sys

sys.setrecursionlimit(100)

def errfactorial(input_nilai):
    print(f"\nproses factorial {input_nilai}")
    try:
        n = int(input_nilai)
        if n < 0:
            raise ValueError("input harus positif")
        hasil = factorial(n)
    except  ValueError as e:
        #print("input harus angka")
        print(f"error: {e}")
    except  RecursionError:
        print("[Error RecursionError] : terlalu besar")
    except  Exception as e:
        #print("input harus angka")
        print(f"error: {e}")
    else:
        print(f"faktorial dari {input_nilai} adalah {hasil} ")
    finally:
        print("proses selesai")

errfactorial(5)          #try->else->finally
errfactorial("lima")     #try->except ValueError->finally
errfactorial(-3)         #try->except ValueError(modif)->finally
errfactorial(150)        #try->except RecursionError->finally
                         #try->except Exception-finally (jika tidak ada RecursionError)
```

```python
def pembagian():
    print("Operasi Pembagian")
    try:
          angka1=float(input("masukan angka 1 : "))
          angka2=float(input("masukan angka 2 : "))
          hasil=angka1/angka2
    except ValueError:
        print("input harus angka")
    except ZeroDivisionError:
        print("pembagi tidak boleh 0")
    else:
        print(f"pembagian dari {angka1}/{angka2} adalah {hasil}")
    finally:
        print("selesai")

pembagian()
```

```python
#class error

class validasiUmur(Exception):
      def __init__(self, message = "usia belum cukup"):
          self.message = message
          super().__init__(f"usia belum cukup  - {self.message}")

def daftarakun(nama,usia):
    print(f"\ndaftar akun untuk {nama} dengan usia {usia} tahun")
    try:
        if usia < 0:
            raise ValueError("usia tidak boleh negatif")
        if usia < 17:
            raise validasiUmur(f"sorry {nama}, usia {usia} tahun kurang dari 17 tahun")
        if usia >= 50:
            raise ValueError("usia sudah uzur")
    except ValueError as e:
        print(f"[ValueError] : {e}")
    except validasiUmur as e:
        print(f"[Error validasiUmur] : {e}")
    else:
        print(f"{nama} umur {usia} tahun pendaftaran Ok")
    finally:
        print("Proses selesai...")

daftarakun("roma",22)     #try->else->finally
daftarakun("wawan",12)    #try->except validasiUmur->finally
daftarakun("cici",-3)     #try->except ValueError->finally
daftarakun("kakek",50)
```
